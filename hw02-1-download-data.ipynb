```python
import os
import boto3

# part 2.1 


# call s3 bucket with data 
s3 = boto3.client('s3')
source_bucket = 'dsan6000-wikipedia'
output_dir = 'data'

# make data folder 
os.makedirs(output_dir, exist_ok=True)

# list files in bucket 
response = s3.list_objects_v2(Bucket=source_bucket)

# iterate through objects 
for obj in response.get('Contents', []):
    file_key = obj['Key']
    
    # check if file is in the right format
    if file_key.endswith('.parquet'):
        local_path = os.path.join(output_dir, os.path.basename(file_key))
        
        print(f"downloading {file_key}...")
        s3.download_file(source_bucket, file_key, local_path)

print("all Parquet data has been downloaded")
```
